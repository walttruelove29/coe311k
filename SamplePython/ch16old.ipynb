```python
import numpy as np
import matplotlib.pyplot as plt
# Euler method function
def euler_method(h, x_start=0, x_end=10):
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1  # y(0) = 1

    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    return x_values, y_values

# Functions representing dy/dx = -2y and exact solution
def f(x, y):
    return -2 * y
def exact_y(x):
    return np.exp(-2*x)

# Error Function
def calculate_errors(x_values, y_values):
    y_exact = exact_y(x_values)
    local_error = np.abs(y_values - y_exact)
    global_error = np.sum(local_error)
    return local_error, global_error
```

```python
# Plotting different values of h vs. exact solution
h_values = [0.1, 0.5, 1.0, 1.5]
x_exact = np.linspace(0, 10, 1000)
y_exact = exact_y(x_exact)
for h in h_values:
    x_values, y_values = euler_method(h)
    plt.plot(x_values, y_values, label= f'Euler\'s Method (h={h})')
plt.xlabel('x')
plt.ylabel('y')
plt.plot(x_exact, y_exact, label='Exact Solution')
plt.title('Exact Solution for dy/dx = -2y vs Euler\'s Method Approximations')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Error plotting
for h in h_values:
    x_values, y_values = euler_method(h)
    local_error, global_error = calculate_errors(x_values, y_values)
    plt.plot(x_values, local_error, label=f'Local Error for h={h}')
plt.xlabel('x')
plt.ylabel('Error')
plt.title('Error for Euler\'s Method with increasing values of h')
plt.legend()
plt.grid(True)
plt.show()
```
