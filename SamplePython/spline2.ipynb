```python
def evaluate_spline(x_data, y_data, n_pts=500):
    a, b, c, d = setup_tridiagonal_matrix(x_data, y_data)
    M = np.array([0.0, *thomas_algorithm(a, b, c, d), 0.0], dtype=float)

    x_eval = np.linspace(x_data[0], x_data[-1], n_pts)
    y_eval = np.zeros(n_pts)

    for k, xv in enumerate(x_eval): 
        # find which interval xv falls in
        i = np.searchsorted(x_data, xv, side='right') - 1
        i = np.clip(i, 0, len(x_data) - 2)   # stay in bounds at right edge

        hi      = x_data[i + 1] - x_data[i]
        dx_r    = x_data[i + 1] - xv          # distance to right knot
        dx_l    = xv - x_data[i]              # distance to left knot

        y_eval[k] = (
            (M[i]     / (6 * hi)) * dx_r**3
          + (M[i + 1] / (6 * hi)) * dx_l**3
          + (y_data[i]     / hi - (M[i]     * hi) / 6) * dx_r
          + (y_data[i + 1] / hi - (M[i + 1] * hi) / 6) * dx_l
        )

    return x_eval, y_eval
```
