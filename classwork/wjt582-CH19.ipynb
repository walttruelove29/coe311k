```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define parameters
h_values = [0.1, 0.01, 0.001]  # step sizes
t_start, t_end = 0, 30

# Function representing dy/dx = -1000y+3000-2000e^-t
def f(t, y):
    return -1000 * y + 3000 -2000 * np.exp(-t)
def analytical_sol(t):
    return 3 - 0.998 * np.exp(-1000*t) - 2.002 * np.exp(-t)

# Implement Euler's method
for h in h_values:
    num_steps = int((t_end - t_start) / h)
    t_values = np.linspace(t_start, t_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_true = analytical_sol(t_values)
    # Initial condition
    y_values[0] = 0  # y(0) = 0
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(t_values[i], y_values[i])
    plt.figure(figsize=(10, 5))
    plt.plot(t_values, y_values, label="Euler's Method Approximation")
    plt.plot(t_values, y_true, label="Analytical Solution")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title("Approximation of dy/dx = -1000y+3000-2000e^-t using Euler's Method")
    plt.legend()
    plt.show()
```
