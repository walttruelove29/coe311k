```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
h = 0.1
# Run with h=0.1 for error calculations
x_start, x_end = 0, 2
num_steps = int((x_end - x_start) / h)
x_values1 = np.linspace(x_start, x_end, num_steps + 1)
y_values1 = np.zeros(num_steps + 1)

# Initial condition
y_values1[0] = 1  # y(0) = 1
# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

# Implement Euler's method
for i in range(num_steps):
    y_values1[i + 1] = y_values1[i] + h * f(x_values1[i], y_values1[i])

# Plot the results
plt.plot(x_values1, y_values1, label="Euler's Method Approximation")
plt.xlabel('x')
plt.ylabel('y')
plt.title(f"Approximation of dy/dx = -2y using Euler's Method with step size {h:.2f}")
plt.legend()
plt.show()
h += 0.05

for i in range(2, 20): # Looping Step Size
    x_start, x_end = 0, 2
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)

# Initial condition
    y_values[0] = 1  # y(0) = 1
# Function representing dy/dx = -2y
    def f(x, y):
        return -2 * y

# Implement Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

# Plot the results
    plt.plot(x_values, y_values, label="Euler's Method Approximation")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f"Approximation of dy/dx = -2y using Euler's Method with step size {h:.2f}")
    plt.legend()
    plt.show()
    h += 0.05
# Error Calculation
    y_valuesinterp = np.interp(x_values1, x_values, y_values) 
    pointerror = np.abs(y_valuesinterp-y_values1)
    print(f"Error Between Individual Points of the First and Current Runs: {np.round(pointerror, 3)}")
    accerror = np.sum(pointerror)
    print(f"Accumulated Error Between the First and Current Runs: {accerror:.3f}")
```

As step size increases, error between the original and current runs increases as well. The error becomes pronounced very quickly, with the total accumulated error exceeding 1 with a step size of only 0.40
