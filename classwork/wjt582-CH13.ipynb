```python
import numpy as np
import time as tm
import matplotlib.pyplot as plt
```

```python
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
```

When comparing these two functions, I expect that the optimized function will consistently be much quicker than the naive one. As for the graphs themselves, I believe that both will exhibit a somewhat similar relationship between time and the size of the matrix, but I believe they will be on different scales.

```python
rng = np.random.default_rng()
n = 3
naivetime = np.zeros(8)
opttime = np.zeros(8)
size = np.zeros(8)
for i in range(8):
    A = rng.uniform(low=1.0, high=10.1, size=(n, n))
    B = rng.uniform(low=1.0, high=10.1, size=(n, n))
    start1 = tm.time()
    C1 = naive_matrix_multiplication(A, B)
    end1 = tm.time()
    time1 = end1 - start1
    start2 = tm.time()
    C2 = np.dot(A, B)
    end2 = tm.time()
    time2 = end2 - start2
    naivetime[i] = time1
    opttime[i] = time2
    size[i] = n
    print(f"Size: {n}x{n}")
    print("Naive Function:")
    
    print(f"Time: {time1}")
    print("Optimized Function:")
    
    print(f"Time: {time2}")
    
    n = n*2
```

```python
print(naivetime)
print(opttime)
print(size)
```

```python
plt.figure()
plt.plot(size, naivetime)
plt.xlabel('NxN size of the Multiplied Matrices')
plt.ylabel('Runtime')
plt.title('Naive Multiplication Matrix Number vs Runtime')
plt.grid()
plt.show()
```

```python
plt.figure()
plt.plot(size, opttime)
plt.xlabel('NxN size of the Multiplied Matrices')
plt.ylabel('Runtime')
plt.title('Optimized Multiplication Matrix Number vs Runtime')
plt.grid()
plt.show()
```

Both graphs ultimately exhibited a similar relationship between matrix number and runtime for both functions, and runtime for the optimized function was consistently faster than the naive one.
