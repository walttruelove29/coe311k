```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points
# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)
```

```python
# Step size
h = (b - a) / (N - 1)
# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)
# Forward Difference Approximation
df_dx1 = (f(x + h) - f(x))/h
# Backward Difference Approximation
df_dx2 = (f(x) - f(x - h))/h
```

```python
# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx, 'o--', label='Central Difference Approximation')
plt.plot(x, df_dx1, '^--', label='Forward Difference Approximation')
plt.plot(x, df_dx2, 'x--', label='Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Central, Forward, and Backward Differences")
plt.legend()
plt.show()
```
